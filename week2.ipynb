#Скачивание библиотек

```python
import pandas as pd
import numpy as np
from pandas import DataFrame
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
```

#Шаг 1 работа с дата сетом

```python
df = fetch_california_housing(as_frame=True).frame #добавила фрейм чтобы сразу был готовый дата сет
```

```python
df.describe() #среднее значение, минимальное, максимальное, стандартное отклонение
```

```python
df.head(7)
```

#Шаг 2 формирую x и y

```python
y=df['MedHouseVal']
x=df.drop('MedHouseVal',axis=1)
```

```python
x.notna().count() #все ячейки заполнены
```

#Шаг 3 делю данные

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=78)
```

```python
print('x_train shape:', x_train.shape)
print('y_train shape:', y_train.shape)
print('x_test shape:', x_test.shape)
print('y_test shape:', y_test.shape)
```

#Шаг 4 обучаю модель

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

#Шаг 5 делаю предсказания

```python
y_pred=model.predict(x_test)
```

```python
df_pred = pd.DataFrame({'Реальная цена': y_test.values,'Предсказание': y_pred})
```

```python
df_pred['Ошибка']=abs(df_pred['Реальная цена']-df_pred['Предсказание']) #df_pred['Ошибка']=abs(y_test-y_pred) не подойдет из за различий в индаксах
```

```python
df_pred
```

#Шаг 6 MAE и RMSE

```python
mae=mean_absolute_error(y_test,y_pred) #просто средняя ошибка
```

```python
RMSE=np.sqrt(mean_squared_error(y_test,y_pred)) #крупные ошибки получают большее отклонение за счет квадрата
```

```python
mae
```

```python
RMSE #больше mae значит есть такие значения где она ошибалась очень сильно
```

```python
y_m=y.mean()
```

```python
print(mae/y_m) #ошибка в 25% является достаточно большой для жизненных задач
```

#Шаг 7 проверяю переобучение

```python
y_trainpr=model.predict(x_train)
```

```python
mae_train=mean_absolute_error(y_train,y_trainpr)
```

```python
mae_train
```

```python
razrav=mae-mae_train
print(razrav) #это хороший показатель которые означанет что модель не переобучилась.она одинаково предсказывает и на тестовых данных и на тренировочных
```

Сильнее всего на стоимость недвижимости влияет количество спален. Это можно объяснить тем, что большее число спален обычно соответствует большей общей площади квартиры, а значит, и более высокой стоимости.

Я отсортировала признаки и выяснила, что в районах с более высоким уровнем дохода недвижимость стоит дороже — это вполне логично. Признаки, значения которых близки к нулю, практически не оказывают влияния на итоговый результат.

Кроме того, модель учитывает географический фактор: как правило, чем южнее расположен объект, тем выше его стоимость.

Ранее мы искали похожие векторы с помощью косинусной близости, а теперь определяем веса посредством скалярного произведения.

Самые важные пункты, которые влияют на цену, – это количество спален и доход района (чем больше, тем дороже), а ещё география – где стоит дом, тоже решает.
